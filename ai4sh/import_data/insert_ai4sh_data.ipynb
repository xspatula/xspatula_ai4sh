# AI4SH in-situ soil data

This notebook is in a customized project for loading data and metadata collected as part of the EU funded project [AI4SoilHealth](https://ai4soilhealth.eu). The project requires customization of the xspatula core framework. By default the path to the start module for a custom project is ./src/"project_name"; it needs to contain a startup function. The default name of that function is "Run_process". 

For detailed instructions on this notebook, please consult the online manual for [Loading AI4SH Data](https://xspatula.github.io/xspatula_ai4sh_docs/loading_data/).

## Utility data

All required utility data are shipped together with the notebook in the `xspatula_ai4sh` GitHub repo. This notebook inserts utility data in the database in a **single-step**. Each cell in the `utility` block translates data straight from excel/csv into the database in one step, using the `insert_tabular_data` process. This happens in 2 steps; first the excel data is translated to JSON command files and then the JSON command files are read and compared with the database process records that tell where each data point should be stored in the database. The intermediate JSON files are stored under sub-directories called `staging` (e.g. ./lucas/import_data/utility/general/insert_process/staging). There is no manual hand-off between steps, and this route is **INSERT-only**: existing records are left untouched, never updated. If you need to UPDATE or DELETE data, or want to inspect/edit the generated JSON before it is applied, please consult the [manual](https://xspatula.github.io). In short you can either manually edit the JSON files under `staging` and write a new notebook cell that points to the edited file and rerun it, or you can use a **two-step** process that first translates the tabular data to JSON process files and then use a second manage command after editing the JSON process file.

## AI4SoilHealth data

AI4SH specific data are not shipped with the notebook but have to be requested. If you want to load the AI4SH eDNA metbarcoding data to the database you must first run the Command Line Interface (CLI) script, `edna_final_results_to_xspatula`, included in the repo as oulined in the online documentation on [Load eDNA Metabarcoding Data](https://xspatula.github.io/xspatula_ai4sh_docs/edna/). The eDNA data is directly transformed into JSON files that are ready for specific xspatula processes to interpret and insert in the database. If you make changes to the database you need to also edit the CLI that generate the JSON files to follow any changes made in the table structure.

## 3.4 AI4SH eDNA data

The AI4SoilHealth environmental DNA (eDNA) data include prokaryotes and fungi core taxonimic groups and functions, and ASV (Amplicon Sequence Variant) taxa tables. 

The results are highly dependent on the methods and e.g. versions of sorftware and reference libraries applied, and must be recorded for full scientific documentation. The method metadata is part of the observation utilities and assembled in the table `observation_utility.method_pipeline` and `observation_utility.method_pipeline_step` that links to each individual method stage that must be registered beforehand in `observation_utility.analysis_method`. The table connections for defining eDNA are illustrated in the database setup manual document [eDNA Metabarcoding](https://xspatula.github.io/xspatula_ai4sh_docs/setup_db/edna_metabarcoding/#how-the-tables-connect).

For AI4SH the actual data results are produced as an excel document with 3 sheets (Summary, ASV_prokaryotes, ASV_fungi), available upon request. To convert the excel document to data that can be read by Xspatula it must be translated with the CLI `edna_final_results_to_xspatula.py` included in the online repo - see the manual page on [Loading eDNA Metabarcoding data](https://xspatula.github.io/xspatula_ai4sh_docs/edna/).

Loading the eDNA data must start with the taxa recorded i the ASV tables, with one cell for loading prokaryote taxa and one for fungi taxa (the source data is generated by the CLI). Then follows the observation log that refers back to the method and the method pipeline; then the summary observations. Lastly the ASV abundances for prokaryotes and fungi are loaded.
