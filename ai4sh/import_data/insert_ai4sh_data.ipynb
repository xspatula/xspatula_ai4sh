## Imports and setup

This notebook is in a customized project and requires a custumized python module (.py file) to run. By default the path to the start module for a custom project is ./src/"project_name"; it needs to contain a startup function. The default name of that function is "Run_process". In this project, [AI4SH](https://ai4soilhealth.eu), the path to the custom project is ./src/ai4sh and the first cell calls the Run_process function in that package.

This notebook is the single-step counterpart to `load_ai4sh_data.ipynb`, the "everything" notebook. Every section that has an excel/csv source (utility, dataset metadata, sample geotagging, Agrolab wetlab) is collapsed from its original translate+manage cell pair(s) into one insert cell, exactly as in the individual `insert_ai4sh_*.ipynb` notebooks. The spectrometer/observation sections (DS2500, Neospectra, FTIR, LIBS) and the LUCAS/OSSL section have no excel/csv source at all - they read hand/script-generated JSON directly - so those cells are carried over unchanged from `load_ai4sh_data.ipynb`; there is nothing for a single-step insert process to collapse there. This route is INSERT-only throughout - use `load_ai4sh_data.ipynb` instead if you need to UPDATE existing records.
